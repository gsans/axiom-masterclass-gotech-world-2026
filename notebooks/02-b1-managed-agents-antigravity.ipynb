## What to Observe
1. **`base_agent="antigravity-preview-05-2026"`**: How Google's Managed Agents runtime provisions an isolated `/workspace/` sandbox (persisted for 7 days).
2. **Multi-Turn Chaining via `previous_interaction_id`**: Enforcing **Slow Engineering** phases (*PLAN -> EXECUTE -> VERIFY*) where each stage is a bounded interaction with `background=True`.


```python
from google import genai
from google.genai import types

client = genai.Client()

# Step 1: Create a Managed Antigravity Agent with an isolated /workspace/ sandbox
managed_agent = client.agents.create(
    model="gemini-3.8-flash",
    base_agent="antigravity-preview-05-2026",
    name="axiom-sandbox-worker",
    instruction=(
        "Follow Slow Engineering Principle 03 (Rules in Code) and Principle 04 (Evidence First). "
        "Write all artifacts to /workspace/ and emit a verification summary before finishing."
    ),
)
print(f"Created Managed Agent: {managed_agent.name} | Base: {managed_agent.base_agent}")
```

```python
# Step 2: Chain Phase 1 (PLAN) -> Phase 2 (EXECUTE & VERIFY) in background mode
plan_interaction = client.interactions.create(
    agent=managed_agent.name,
    input="Draft an architecture spec in /workspace/SPEC.md for an MCP weather tool server.",
    background=True,
)
print(f"[PHASE 1: PLAN] Interaction ID: {plan_interaction.id} | Status: COMPLETED")

verify_interaction = client.interactions.create(
    agent=managed_agent.name,
    previous_interaction_id=plan_interaction.id,
    input="Verify /workspace/SPEC.md against Principle 10 (strict JSON schema input gates) and emit DecisionCard.",
    background=True,
)
print(f"[PHASE 2: VERIFY] Chained to {plan_interaction.id} -> Output: /workspace/DECISION_CARD.json")
```
