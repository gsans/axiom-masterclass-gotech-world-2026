## What to Observe
1. **Exposing an ADK Agent via `to_a2a()`**: Auto-generates the standard `/.well-known/agent-card.json` endpoint declaring skills, capabilities, and streaming support.
2. **Consuming a Peer Swarm via `RemoteA2aAgent`**: How a coordinator agent discovers a remote specialist's Agent Card and delegates sub-tasks over JSON-RPC / SSE.


```python
from google.adk.agents import Agent
from google.adk.agents.remote_a2a_agent import RemoteA2aAgent, AGENT_CARD_WELL_KNOWN_PATH
from google.adk.a2a.utils.agent_to_a2a import to_a2a

# 1. Specialist Agent exposed over A2A (Port 4100)
specialist_agent = Agent(
    name="axiom_research_specialist",
    model="gemini-4-argon",
    description="Remote A2A agent specializing in deep specification verification and Decision Card drafting.",
    instruction="Validate incoming architecture specs against Slow Engineering rules and return a DecisionCard.",
)

# Expose FastAPI/Starlette A2A server (serves /.well-known/agent-card.json automatically)
a2a_app = to_a2a(specialist_agent, port=4100)
print(f"Exposed A2A Specialist at http://localhost:4100{AGENT_CARD_WELL_KNOWN_PATH}")

# 2. Coordinator Agent consuming the Remote A2A Specialist
remote_specialist = RemoteA2aAgent(
    name="remote_research_specialist",
    description="Delegates deep verification and ticket creation to the remote A2A specialist.",
    agent_card=f"http://localhost:4100{AGENT_CARD_WELL_KNOWN_PATH}",
)

coordinator_agent = Agent(
    name="mission_control_coordinator",
    model="gemini-3.8-flash",
    instruction="Route specification audits and Decision Card generation to remote_research_specialist.",
    sub_agents=[remote_specialist],
)
print("Wired Coordinator -> RemoteA2aAgent handshake.")
```
