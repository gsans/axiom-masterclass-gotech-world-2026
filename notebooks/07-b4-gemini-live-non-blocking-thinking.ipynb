## How This Powers the Personal AI Heartbeat Loop (Slow Engineering Principle 05)
When you speak a long-horizon intent to your Personal AI (*"Research the A2A spec, audit our MCP server, and open a ticket"*):
1. **Non-Blocking Tool Declaration (`behavior="NON_BLOCKING"`)**: The Gemini Live voice session immediately acknowledges your voice request conversationally without freezing the audio stream.
2. **Background Swarm Dispatch (`spawn_background_task`)**: The tool triggers your ADK + MCP + A2A agent team in the background while the Angular UI displays a pulsing **Heartbeat Status (`IN_PROGRESS` -> `DECISION_CARD_READY`)** using Signals.
3. **Human Beat Notification**: Once the ADK verification loop finishes, the completed `DecisionCard` is pushed back into the Live session with `scheduling="WHEN_IDLE"`, so your Personal AI briefs you cleanly without interrupting you mid-sentence.


```python
from google import genai
from google.genai import types

# Declare the Non-Blocking Voice-to-Swarm Bridge Tool
spawn_task_declaration = types.FunctionDeclaration(
    name="spawn_background_task",
    description=(
        "Spawns a long-horizon background job on the ADK + MCP agent team "
        "(research, code generation, specification audit, or ticket creation)."
    ),
    parameters=types.Schema(
        type="OBJECT",
        properties={
            "intent": types.Schema(type="STRING", description="Clear description of the engineering job."),
            "task_type": types.Schema(
                type="STRING",
                enum=["RESEARCH", "CODE_GEN", "SPEC_AUDIT", "TICKET"],
            ),
        },
        required=["intent", "task_type"],
    ),
    behavior="NON_BLOCKING",  # Voice conversation continues smoothly while swarm works!
)

capstone_live_config = types.LiveConnectConfig(
    response_modalities=[types.Modality.AUDIO],
    tools=[types.Tool(function_declarations=[spawn_task_declaration])],
)

# When the background ADK + MCP swarm finishes its bounded loop, return the Decision Card WHEN_IDLE:
completed_decision_card_response = types.FunctionResponse(
    name="spawn_background_task",
    id="call_heartbeat_01",
    response={
        "interaction_status": "IDLE",
        "decision_card": {
            "jobId": "job-live-01",
            "whatWasDone": "Audited MCP server against Principle 10 and drafted ticket ENG-402.",
            "checksPassed": True,
            "questionForHuman": "Approve publishing ticket ENG-402 to GitHub Issues?",
        },
        "scheduling": "WHEN_IDLE",  # Respects Slow Engineering Principle 05: The Human Beat
    },
)

print("Configured NON_BLOCKING Gemini Live voice-to-swarm bridge (WHEN_IDLE human beat delivery).")
```
